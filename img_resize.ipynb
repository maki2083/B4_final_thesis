```python
import cv2
import os

def resize_image_for_analysis(input_path, max_size=1500):
    """
    画像を解析に適したサイズ（長辺がmax_sizeになるよう）にリサイズして上書き保存、
    またはリサイズ版のパスを返す関数。
    今回は安全のため、'_resized' という別名で保存します。
    """
    if not os.path.exists(input_path):
        print(f"エラー: 画像が見つかりません {input_path}")
        return None

    img = cv2.imread(input_path)
    h, w = img.shape[:2]
    
    # すでに十分小さいなら何もしない
    if max(h, w) <= max_size:
        print(f"画像サイズは適切です ({w}x{h})。リサイズは不要です。")
        return input_path

    # リサイズ比率の計算
    scale = max_size / max(h, w)
    new_w, new_h = int(w * scale), int(h * scale)
    
    resized_img = cv2.resize(img, (new_w, new_h), interpolation=cv2.INTER_AREA)
    
    # ファイル名を変更して保存 (例: test.jpg -> test_resized.jpg)
    root, ext = os.path.splitext(input_path)
    output_path = f"{root}_resized{ext}"
    
    cv2.imwrite(output_path, resized_img)
    print(f"画像をリサイズしました: {w}x{h} -> {new_w}x{new_h}")
    print(f"新しい画像パス: {output_path}")
    
    return output_path

# --- 使い方 ---
# 元の画像パス
original_image_path = "images/DJI_0471.JPG" 

# リサイズを実行し、新しいパスを取得
# この new_image_path を、以降の image_path 変数に設定して使ってください
new_image_path = resize_image_for_analysis(original_image_path, max_size=1200)

# ↓ これをパラメータ設定に反映させるのを忘れずに！
# image_path = new_image_path
```
